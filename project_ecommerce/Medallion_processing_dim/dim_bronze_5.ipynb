# Bronze Ingestion — Date

Loads the raw `date.csv` source data from the e-commerce source volume into `ecommerce.bronze.bronze_date`, keeping raw values untouched. Two metadata columns are added:

* `source_file` — the full path of the source file the row came from
* `ingested_at` — the timestamp when the row was ingested

```python
from pyspark.sql.functions import col, current_timestamp

source_path = "/Volumes/ecommerce/source_data/raw_data/ecomm-raw-data/date/"
target_table = "ecommerce.bronze.bronze_date"

# Read the raw CSV as-is (header row present) from the date directory
df_date_raw = spark.read.csv(source_path, header=True, inferSchema=True)

# Attach ingestion metadata: full source file path and ingestion timestamp
df_bronze_date = df_date_raw.withColumn(
    "source_file", col("_metadata.file_path")
).withColumn(
    "ingested_at", current_timestamp()
)

# Idempotent full load into the bronze table
(
    df_bronze_date.write
    .mode("overwrite")
    .saveAsTable(target_table)
)

print(f"Wrote {spark.table(target_table).count()} rows to {target_table}")
```
